# Hai đồ thị hồi quy Logistic - Bài 2

Notebook vẽ đồ thị câu **h** (dữ liệu, Sigmoid và ranh giới quyết định sau 2 lần cập nhật) và câu **j** (loss qua 1.000 lần cập nhật).

Chạy lần lượt các ô từ trên xuống. Có thể sửa dữ liệu, tốc độ học hoặc tham số trong ô đầu để vẽ lại.

```python
import numpy as np
import matplotlib.pyplot as plt

# 14 mau du lieu trong de: thoi luong (phut), nhan G=1 va B=0
x_phut = np.array([132, 141, 88, 154, 151, 102, 119, 106, 102, 143, 76, 168, 94, 186.], dtype=float)
y = np.array([1, 1, 0, 1, 1, 0, 0, 0, 0, 1, 0, 1, 0, 1.], dtype=float)

# Chuan hoa dac trung nhu trong de: x = (thoi luong - 130) / 50
x = (x_phut - 130) / 50
alpha = 0.2
w0, b0 = 0.0, 0.0

# Font serif va math gan voi phong cach cong thuc trong tai lieu
plt.rcParams['font.family'] = 'DejaVu Serif'
plt.rcParams['mathtext.fontset'] = 'stix'
plt.rcParams['axes.unicode_minus'] = False

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def loss_value(w, b):
    p = np.clip(sigmoid(w*x + b), 1e-12, 1 - 1e-12)
    return -np.mean(y*np.log(p) + (1-y)*np.log(1-p))

def train_steps(w, b, n):
    for _ in range(n):
        p = sigmoid(w*x + b)
        gw = np.mean((p-y)*x)
        gb = np.mean(p-y)
        w, b = w-alpha*gw, b-alpha*gb
    return w, b
```

## Đồ thị 1 - Câu h: Sigmoid và ranh giới quyết định

Mô hình dùng tham số sau 2 lần Batch Gradient Descent. Ranh giới tại ngưỡng 0.5 thỏa `w*x + b = 0`.

```python
w2, b2 = train_steps(w0, b0, 2)
x_line_phut = np.linspace(x_phut.min()-5, x_phut.max()+5, 500)
x_line = (x_line_phut-130)/50
p_line = sigmoid(w2*x_line+b2)
boundary_phut = 130 + 50*(-b2/w2)

fig, ax = plt.subplots(figsize=(9, 4.8))
ax.scatter(x_phut[y==0], y[y==0], color="black", s=34, label="B (0)", zorder=3)
ax.scatter(x_phut[y==1], y[y==1], color="red", s=34, label="G (1)", zorder=3)
ax.plot(x_line_phut, p_line, color="blue", linewidth=1.8, label="Sigmoid")
ax.axhline(0.5, color="gray", linestyle="--", linewidth=0.9)
ax.axvline(boundary_phut, color="gray", linestyle=":", linewidth=1.2, label=f"Ranh giới: {boundary_phut:.1f} phút")
ax.set(xlabel="Thời lượng (phút)", ylabel="Xác suất / nhãn", xlim=(70,190), ylim=(-0.08,1.08), yticks=[0,0.5,1])
ax.legend(frameon=False, ncol=2)
ax.spines[["top","right"]].set_visible(False)
ax.grid(axis="y", color="0.9", linewidth=0.7)
fig.tight_layout()
plt.show()
print(f"Sau 2 lần cập nhật: w={w2:.6f}, b={b2:.6f}; ranh giới ≈ {boundary_phut:.3f} phút")
```

## Đồ thị 2 - Câu j: Hàm mất mát theo số lần lặp

Lưu loss ở trạng thái ban đầu và sau mỗi lần cập nhật, để trục hoành chạy từ 0 đến 1.000. Đường ngang chấm thể hiện loss sau 2 lần cập nhật.

```python
n_steps = 1000
w, b = w0, b0
losses = [loss_value(w, b)]
for _ in range(n_steps):
    p = sigmoid(w*x+b)
    gw = np.mean((p-y)*x)
    gb = np.mean(p-y)
    w, b = w-alpha*gw, b-alpha*gb
    losses.append(loss_value(w, b))

loss2 = loss_value(w2, b2)
fig, ax = plt.subplots(figsize=(9, 4.8))
ax.plot(np.arange(n_steps+1), losses, color="blue", linewidth=1.8, label="Loss huấn luyện")
ax.axhline(loss2, color="red", linestyle="--", linewidth=1.1, label=f"Sau 2 lần cập nhật: {loss2:.4f}")
ax.set(xlabel="Số lần cập nhật", ylabel="Binary Cross-Entropy Loss", xlim=(0,n_steps))
ax.spines[["top","right"]].set_visible(False)
ax.grid(color="0.9", linewidth=0.7)
ax.legend(frameon=False)
fig.tight_layout()
plt.show()
print(f"Loss ban đầu: {losses[0]:.6f}")
print(f"Loss sau 2 lần cập nhật: {loss2:.6f}")
print(f"Loss sau 1.000 lần cập nhật: {losses[-1]:.6f}")
print(f"Tham số sau 1.000 lần cập nhật: w={w:.6f}, b={b:.6f}")
```
