# P2 - Câu 1 và Câu 2: Miền kích hoạt

Notebook tạo hai hình riêng. Chạy các ô theo thứ tự từ trên xuống.

Theo vector trọng số đầu ra trong đề, $v_4=2o_1+3o_2+o_3+1\geq1$ với mọi $(x_1,x_2)$. Vì vậy, miền đầu ra của toàn mạng là toàn mặt phẳng, cả trước và sau khi đổi bias. Các đường màu là ranh giới kích hoạt của tầng ẩn.

```python
import numpy as np
import matplotlib.pyplot as plt

# Plot window; edit these limits to view a wider or narrower region.
x1_min, x1_max = -2.0, 3.0
x2_min, x2_max = -1.5, 2.5
x1 = np.linspace(x1_min, x1_max, 500)
x2 = np.linspace(x2_min, x2_max, 500)
X1, X2 = np.meshgrid(x1, x2)

def step(z):
    return (z >= 0).astype(int)

def draw_activation_figure(hidden_bias, title, filename, color):
    # Weighted sums at the three hidden neurons.
    v1 = 2*X1 + hidden_bias[0]
    v2 = X1 + 2*X2 + hidden_bias[1]
    v3 = X1 + 3*X2 + hidden_bias[2]
    o1, o2, o3 = step(v1), step(v2), step(v3)

    # Output neuron weights are [2, 3, 1, 1], including output bias +1.
    v4 = 2*o1 + 3*o2 + o3 + 1
    y = step(v4)

    fig, ax = plt.subplots(figsize=(7.2, 4.8))
    ax.contourf(X1, X2, y, levels=[-0.5, 0.5, 1.5],
                colors=["#e6f0fa", "#e6f0fa"], alpha=0.95)
    xx = np.linspace(x1_min, x1_max, 500)

    # Hidden neuron 1: 2*x1 + bias = 0.
    x1_boundary = -hidden_bias[0] / 2
    ax.axvline(x1_boundary, color=color, linewidth=1.8,
               label=f"h1: x1 = {x1_boundary:g}")
    # Hidden neuron 2: x1 + 2*x2 + bias = 0.
    ax.plot(xx, (-xx-hidden_bias[1])/2, color="#d27a28", linewidth=1.8,
            label="h2 boundary")
    # Hidden neuron 3 keeps bias 0 in both questions.
    ax.plot(xx, -xx/3, color="#56843b", linestyle="--", linewidth=1.8,
            label="h3 boundary (unchanged)")

    ax.set(xlim=(x1_min, x1_max), ylim=(x2_min, x2_max),
           xlabel="x1", ylabel="x2", title=title)
    ax.text(0.03, 0.96, "Network output y = 1 everywhere",
            transform=ax.transAxes, va="top", fontsize=9,
            bbox={"facecolor":"white", "alpha":0.88, "edgecolor":"none"})
    ax.grid(alpha=0.22)
    ax.legend(loc="lower left", frameon=True)
    fig.tight_layout()
    fig.savefig(filename, dpi=200, bbox_inches="tight")
    plt.show()
    print("Saved:", filename, "| unique output:", np.unique(y))
    plt.close(fig)
```

## Hình 1 - Câu 1: Bias ẩn ban đầu $+1$

Bias ẩn theo thứ tự là $(+1,+1,0)$.

```python
draw_activation_figure((1, 1, 0), "C\u00e2u 1 - Original hidden biases",
                       "P2_C1_mien_kich_hoat.png", "#3679a8")
```

## Hình 2 - Câu 2: Bias ẩn sau khi thay đổi

Hai bias đầu đổi từ $+1$ thành $-1$; bias thứ ba giữ nguyên $0$.

```python
draw_activation_figure((-1, -1, 0), "C\u00e2u 2 - Changed hidden biases",
                       "P2_C2_mien_kich_hoat.png", "#c94f45")
```
